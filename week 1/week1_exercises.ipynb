# Week 1 - Runner Energy Consumption

In this exercise, I will create and plot a function that shows how a runner's energy consumption changes with speed.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
%pip install matplotlib numpy
```

```python
def C(s):
    return 5*s + 0.2*s**3
```

```python
C(5)
```

```python
s_values = np.linspace(0, 10, 500)
```

```python
C_values = C(s_values)
```

```python
plt.plot(s_values, C_values, label="C(s) = 5s + 0.2s³")

plt.title("Runner Energy Consumption")
plt.xlabel("Speed (m/s)")
plt.ylabel("Energy Consumption (calories)")

plt.xlim([0, 10])
plt.ylim([0, 260])

plt.grid(True)
plt.legend(loc="lower right")

plt.show()
```

## Answer

The energy expenditure function is:

C(s) = 5s + 0.2s³

At a speed of 5 m/s:

C(5) = 5(5) + 0.2(5³) = 50

Therefore, the runner uses **50 calories** at a speed of 5 m/s.
